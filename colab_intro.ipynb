## Sumowanie ciągu geometrycznego

Dla ciągu geometrycznego o pierwszym wyrazie $a$ i ilorazie $r$, suma pierwszych $n$ wyrazów wyraża się wzorem:
$$S_n = a \frac{1-r^n}{1-r}, \quad r \neq 1$$

Przykładowa implementacja w języku Python:
```python
def suma_geometryczna(a, r, n):
    if r == 1:
        return a * n
    return a * (1 - r**n) / (1 - r)
```

````python
# 7 & 8. Wykres y = k^2 dla k = 1..20 oraz zapis do wykres.png
import matplotlib.pyplot as plt

k = list(range(1, 21))
y = [val**2 for val in k]

plt.figure(figsize=(8, 5))
plt.plot(k, y, marker='o', linestyle='-', color='b', label='$y = k^2$')
plt.title('Wykres $y = k^2$ dla $k \in [1, 20]$')
plt.xlabel('k')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.savefig('wykres.png', dpi=150)
plt.show()
````

````python
# 9. Animacja wahadła i zapis do video.mp4
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML

# Parametry fizyczne
g = 9.81
L = 1.0
t = np.linspace(0, 10, 200)
theta = 0.8 * np.cos(np.sqrt(g / L) * t)  # proste wahadło

x = L * np.sin(theta)
y = -L * np.cos(theta)

fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 0.2)
ax.set_aspect('equal')
ax.grid(True)

line, = ax.plot([], [], 'o-', lw=2, color='navy')
trace, = ax.plot([], [], '-', lw=1, color='lightblue', alpha=0.7)

def init():
    line.set_data([], [])
    trace.set_data([], [])
    return line, trace

def animate(i):
    line.set_data([0, x[i]], [0, y[i]])
    trace.set_data(x[:i], y[:i])
    return line, trace

anim = animation.FuncAnimation(fig, animate, init_func=init, frames=len(t), interval=50, blit=True)

# Zapisujemy do video.mp4 (wymaga ffmpeg w systemie)
writer = animation.FFMpegWriter(fps=20, metadata=dict(artist='Colab'), bitrate=1800)
anim.save('video.mp4', writer=writer)
plt.close(fig)

# Osadzenie wideo w notebooku
from IPython.display import Video
Video('video.mp4', embed=True, width=400)
````

````python
# 10. Konwersja video.mp4 na video.gif
import imageio

reader = imageio.get_reader('video.mp4')
fps = reader.get_meta_data().get('fps', 20)

writer = imageio.get_writer('video.gif', fps=fps)
for frame in reader:
    writer.append_data(frame)
writer.close()
print("Konwersja zakończona. Animacja została zapisana do pliku video.gif.")
````

````python
# 11. Zmodyfikowana krzywa parametryczna (Lissajous 3D) i zapis do parametric_curve.png
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

t = np.linspace(0, 2 * np.pi, 1000)

# Zmodyfikowane równania parametryczne (Krzywa Lissajous 3D)
x = np.sin(3 * t)
y = np.cos(4 * t)
z = np.sin(5 * t)

fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, label="Krzywa Lissajous 3D", color='purple', lw=2)
ax.set_title("Zmodyfikowana Krzywa Parametryczna (3D)", fontsize=14)
ax.set_xlabel("x(t) = sin(3t)")
ax.set_ylabel("y(t) = cos(4t)")
ax.set_zlabel("z(t) = sin(5t)")
ax.legend()

plt.savefig('parametric_curve.png', dpi=150)
plt.show()
print("Krzywa została zapisana do pliku parametric_curve.png.")
````
